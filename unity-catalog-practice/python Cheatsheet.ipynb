```python
nums = [1,2,3]
nums.append(4)
nums.remove(2)
nums[2]

# This cell demonstrates Python list operations:

# nums = [1,2,3] - Creates a list containing three integers
# nums.append(4) - Adds 4 to the end of the list, making it [1,2,3,4]
# nums.remove(2) - Removes the first occurrence of value 2, resulting in [1,3,4]
# nums[0] - Accesses the first element (index 0) of the list, which returns 1
# The output shows 1 because that's the last expression evaluated - the first element of the modified list after the append and remove operations.

# Lists in Python are mutable (can be changed), ordered collections that can hold items of any type. They support indexing (starting at 0), slicing, and various methods like append(), remove(), insert(), pop(), and more.
```

```python
data = (1, 2, 3)
data[2]

# This cell demonstrates Python tuples:

# data = (1, 2, 3) - Creates a tuple containing three integers using parentheses
# data[0] - Accesses the first element (index 0), which returns 1
# Key difference from lists: Tuples are immutable, meaning once created, you cannot modify them. You can't use methods like append(), remove(), or change individual elements. If you tried data[0] = 5, it would raise an error.

# When to use tuples:

# When you need data that shouldn't change (coordinates, RGB colors, database records)
# As dictionary keys (lists can't be keys because they're mutable)
# Slightly faster and more memory-efficient than lists
# To signal intent that the data is fixed
# Tuples support indexing and slicing just like lists, but lack modification methods.
```

```python
items = {1, 2, 3}
items.add(4)
items.remove(2)

# This cell demonstrates Python sets:

# items = {1, 2, 3} - Creates a set containing three integers using curly braces
# items.add(4) - Adds 4 to the set, making it {1, 2, 3, 4}
# items.remove(2) - Removes 2 from the set, resulting in {1, 3, 4}
# Key characteristics of sets:

# Unordered - No indexing like items[0]; elements have no fixed position
# Unique elements only - Duplicates are automatically removed
# Mutable - You can add/remove elements after creation
# Fast membership testing - Checking if an item exists is very efficient
# Common use cases:

# Removing duplicates from a list: set([1, 2, 2, 3]) → {1, 2, 3}
# Mathematical operations: union (|), intersection (&), difference (-)
# Fast lookups to check if an item exists
# The cell produces no visible output because the operations modify the set in place without returning a value.
```

```python
user = {"name": "Alex", "age": 30}
user["name"]
user.keys()
user.values()

# This cell demonstrates Python dictionaries:

# user = {"name": "Alex", "age": 30} - Creates a dictionary with key-value pairs using curly braces
# user["name"] - Accesses the value associated with key "name", returns "Alex"
# user.keys() - Returns all keys in the dictionary: dict_keys(['name', 'age'])
# user.values() - Returns all values in the dictionary: dict_values(['Alex', 30]) (shown in output)
# Key characteristics of dictionaries:

# Key-value pairs - Store data as associations between unique keys and their values
# Mutable - You can add, modify, or remove entries: user["email"] = "alex@example.com"
# Fast lookups - Retrieving values by key is very efficient
# Keys must be immutable - Strings, numbers, tuples work; lists don't
# Common use cases:

# Storing structured data (user profiles, configuration settings)
# Counting occurrences: {"apple": 3, "banana": 2}
# Mapping relationships between data
# Other useful methods include .get(), .items(), .pop(), and .update().

 
```

```python

# Arithmetic operations
print(10 + 5)  # Addition
print(10 - 5)  # Subtraction
print(10 * 5)  # Multiplication
print(10 / 5)  # Division
print(10 % 5)  # Modulo (remainder)
print(10 ** 5) # Exponentiation (power)
print(10 // 5) # Floor division (integer division)

# Comparison operators
print(10 == 5)  # Equality
print(10 != 5)  # Inequality
print(10 > 5)   # Greater than
print(10 < 5)   # Less than
print(10 >= 5)  # Greater than or equal to
print(10 <= 5)  # Less than or equal to

# Logical operators
print(True and False)  # Logical AND
print(True or False)   # Logical OR
print(not True)        # Logical NOT

# Bitwise operators
print(10 & 5)  # Bitwise AND
print(10 | 5)  # Bitwise OR
print(10 ^ 5)  # Bitwise XOR
print(~10)     # Bitwise NOT
print(10 << 5) # Bitwise left shift
print(10 >> 5) # Bitwise right shift

# Identity operators
print(10 is 5)   # Identity comparison
print(10 is not 5) # Inverse identity comparison

# Membership operators
print(5 in [1, 2, 3])  # Membership check
print(5 not in [1, 2, 3]) # Inverse membership check

# Operator precedence
print(10 + 5 * 2)  # Multiplication has higher precedence than addition
print((10 + 5) * 2) # Parentheses override operator precedence


# Logical operators have higher precedence than comparison operators
print(True and 10 > 5)  # True and True is True
print(True or 10 > 5)   # True or True is True
print(not True and 10 > 5)  # False and True is False
print(not True or 10 > 5)   # False or True is True

# Bitwise operators have higher precedence than logical operators
print(10 & 5 or True)  # 0 or True is True
print(10 & (5 or True)) # 10 & 1 is 2
print((10 & 5) or True) # 0 or True is True
print((10 & 5) or False) # 0 or False is False
print(10 & 5 and True)  # 0 and True is False
print(10 & (5 and True)) # 10 & 1 is 2
print((10 & 5) and True) # 0 and True is False
print((10 & 5) and False) # 0 and False is False

#   Operator precedence
#   1. Parentheses
#   2. Exponentiation
#   3. Multiplication, Division, Modulo, Floor Division
#   4. Addition, Subtraction
#   5. Bitwise Shifts
#   6. Bitwise AND
#   7. Bitwise XOR  
#   8. Bitwise OR
```
